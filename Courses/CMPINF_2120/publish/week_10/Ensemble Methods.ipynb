## Ensemble Methods
### This is the notebook for the Ensemble Methods project

```python
import numpy as np

import pandas as pd

from sklearn.model_selection import train_test_split

from sklearn.neighbors import KNeighborsClassifier

from sklearn.tree import DecisionTreeClassifier

from sklearn.ensemble import VotingClassifier

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, ConfusionMatrixDisplay

from sklearn.linear_model import LinearRegression

from sklearn.neighbors import KNeighborsRegressor

from sklearn.tree import DecisionTreeRegressor

from sklearn.ensemble import VotingRegressor

from sklearn.metrics import mean_squared_error

from sklearn.ensemble import RandomForestClassifier
```

### Load Data

```python
data1 = pd.read_csv("APMM10-clf-dataset02.csv")
```

```python
data1
```

```python
X = data1[["X1", "X2"]]
```

```python
y = data1.y.ravel()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)
```

```python
# KNeighborsClassifier() with K=3
knn3 = KNeighborsClassifier(n_neighbors=3)
# KNeighborsClassifier() with K=4
knn4 = KNeighborsClassifier(n_neighbors=4)
# KNeighborsClassifier() with K=5
knn5 = KNeighborsClassifier(n_neighbors=5)
# DecisionTreeClassifier() with max_depth=3
dct3 = DecisionTreeClassifier(max_depth=3)
# DecisionTreeClassifier() with max_depth=4
dct4 = DecisionTreeClassifier(max_depth=4)
```

```python
voting_clf = VotingClassifier(
    estimators=[('knn3', knn3), ('knn4', knn4), ('knn5', knn5),
               ('dct3', dct3), ('dct4', dct4)],
    voting='hard'  # or 'soft' if you want to average predicted probabilities
)
```

```python
voting_clf.fit(X_train, y_train)
```

```python
y_pred = voting_clf.predict(X_test)
```

```python
accuracy = accuracy_score(y_pred, y_test)
```

```python
accuracy
```

```python
f1 = f1_score(y_test, y_pred)
```

```python
f1
```

```python
for name, vc in voting_clf.named_estimators_.items():
    y_pred = vc.predict(X_test)
    
    # Calculate precision
    precision = precision_score(y_test, y_pred)  # or 'macro'/'micro' if you prefer
    recall = recall_score(y_test, y_pred)
    print(f"{name}: Precision Score = {precision:.4f}")
    print(f"{name}: Recall Score = {recall:.4f}")
```

### Load Dataset 3

```python
data2 = pd.read_csv("APMM10-reg-dataset03.csv")
```

```python
data2.head()
```

```python
# LinearRegression()
lr = LinearRegression()
# KNeighborsRegressor() with K=5
knr5 = KNeighborsRegressor(n_neighbors=5)
# DecisionTreeRegressor() with max_depth=5
dtr5 = DecisionTreeRegressor(max_depth=5)
```

```python
voting_reg = VotingRegressor(
        estimators=[('lr', lr), ('knr5', knr5), ('dtr5', dtr5)])
```

```python
X2 = data2.drop(columns=["y"]).copy()
```

```python
y2 = data2.y.ravel()
```

```python
X_train2, X_test2, y_train2, y_test2 = train_test_split(X2, y2, test_size=0.25, random_state=42)
```

```python
voting_reg.fit(X_train2, y_train2)
```

```python
y_pred2 = voting_reg.predict(X_test2)
```

```python
rsme = np.sqrt(mean_squared_error(y_pred2, y_test2))
```

```python
rsme
```

```python
for name, vc in voting_reg.named_estimators_.items():
    y_pred2 = vc.predict(X_test2)

    rsme = np.sqrt(mean_squared_error(y_pred2, y_test2))
    print(f"{name}: RSME Value: {rsme}")
```

### Load Dataset

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import StackingClassifier
```

```python
data5 = pd.read_csv("APMM10-clf-dataset08.csv")
```

```python
data5.y.value_counts()
```

```python
X5 = data5.drop(columns=["y"]).copy()
```

```python
y5 = data5.y.ravel()
```

```python
X_train5, X_test5, y_train5, y_test5 = train_test_split(X5, y5, test_size=.30, random_state=42, stratify=y5)
```

```python
# LogisticRegression() with a max_iter limit of 1000,
lr = LogisticRegression(max_iter=1000)
# DecisionTreeClassifier() with a max_depth of 3 (and random_state=42),
dtc3 = DecisionTreeClassifier(max_depth=3, random_state=42)
# DecisionTreeClassifier() with a max_depth of 5 (and random_state=42)
dtc5 = DecisionTreeClassifier(max_depth=5, random_state=42)
# KNeighborsClassifier() with k=5, and
knn5 = KNeighborsClassifier(n_neighbors=5)
# KNeighborsClassifier() with k=7
knn7 = KNeighborsClassifier(n_neighbors=7)
```

```python
stack = StackingClassifier(
            estimators=[('dtc3', dtc3), ('dtc5', dtc5), ('knn5', knn5), ('knn7', knn7), ('lr', lr)],
            final_estimator=LogisticRegression(),
            cv=5
)
```

```python
stack.fit(X_train5, y_train5)
```

```python
y_pred5 = stack.predict(X_test5)
```

```python
accuracy5 = accuracy_score(y_test5, y_pred5)
```

```python
accuracy5
```

```python
f1s5 = f1_score(y_test5, y_pred5, average='weighted')
```

```python
f1s5
```

```python
for name, vc in stack.named_estimators_.items():
    y_pred5 = vc.predict(X_test5)
    
    # Calculate precision
    precision = precision_score(y_test5, y_pred5, average='weighted')  # or 'macro'/'micro' if you prefer
    recall = recall_score(y_test5, y_pred5, average='weighted')
    print(f"{name}: Precision Score = {precision:.4f}")
    print(f"{name}: Recall Score = {recall:.4f}")
```
