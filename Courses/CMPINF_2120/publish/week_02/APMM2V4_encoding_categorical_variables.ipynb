```python
# Imports
import pandas as pd
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
import numpy as np
```

```python
# Sample dataset
data = pd.DataFrame({
    'education': ['High School', 'Bachelors', 'Masters', 'PhD', 'Bachelors'],
    'gender': ['Male', 'Female', 'Female', 'Male', 'Female'],
    'income': [40000, 50000, 60000, 80000, 52000]
})
data
```

## Label Encoding (for ordinal data)

```python
# Ordinal encoding for education
education_order = {'High School': 0, 'Bachelors': 1, 'Masters': 2, 'PhD': 3}
data['education_encoded'] = data['education'].map(education_order)
data
```

## One-Hot Encoding (for nominal data)

```python
# One-hot encoding gender
data_encoded = pd.get_dummies(data, columns=['gender'])
data_encoded
```

## Manual encoding for ordinal data

```python
df = pd.DataFrame({
    'education': ['High School', 'Bachelors', 'Masters', 'PhD', 'Bachelors']
})

# Define manual mapping
education_map = {
    'High School': 0,
    'Bachelors': 1,
    'Masters': 2,
    'PhD': 3
}

df['education_encoded'] = df['education'].map(education_map)
df
```

## Manual encoding for nominal data

```python
df = pd.DataFrame({
    'gender': ['Male', 'Female', 'Female', 'Male']
})

df['gender_male'] = (df['gender'] == 'Male').astype(int)
df['gender_female'] = (df['gender'] == 'Female').astype(int)
df
```
