```python
# Imports
import pandas as pd
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler
import numpy as np
```

```python
# Sample dataset - 5 items
data5 = pd.DataFrame({
    'age': [25, 35, 45, 55, 65],
    'income': [40000, 50000, 60000, 80000, 120000]
})
data5
```

```python
# Generate Data Statistics
data5.describe()
```

```python
# Sample dataset - 10 items
data10 = pd.DataFrame({
    'age': [22, 28, 35, 40, 45, 50, 55, 60, 65, 70],
    'income': [35000, 42000, 50000, 55000, 60000, 68000, 80000, 95000, 120000, 140000]
})
data10
```

```python
# Generate Data Statistics
data10.describe()
```

```python
# Pick a dataset (either data5 or data10)

#data = data5
data = data10
```

## Standardization (Z-score scaling)

```python
scaler = StandardScaler()
data_standardized = pd.DataFrame(scaler.fit_transform(data), columns=data.columns)
data_standardized
```

```python
# Generate Data Statistics
data_standardized.describe()
```

## Min-Max Scaling

```python
scaler = MinMaxScaler()
data_minmax = pd.DataFrame(scaler.fit_transform(data), columns=data.columns)
data_minmax
```

```python
# Generate Data Statistics
data_minmax.describe()
```

## Robust Scaling (less sensitive to outliers)

```python
scaler = RobustScaler()
data_robust = pd.DataFrame(scaler.fit_transform(data), columns=data.columns)
data_robust
```

```python
# Generate Data Statistics
data_robust.describe()
```
