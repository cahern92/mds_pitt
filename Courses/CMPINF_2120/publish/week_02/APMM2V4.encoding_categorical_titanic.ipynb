```python
# Imports
import pandas as pd
import seaborn as sns
from sklearn.preprocessing import LabelEncoder, OrdinalEncoder, OneHotEncoder
```

## Load Sample Titanic Data

```python
# Load the dataset from seaborn
df = sns.load_dataset('titanic')
# more info at https://www.kaggle.com/c/titanic/data
df = df[['survived', 'sex', 'embarked', 'class', 'who']].dropna()
df.head(8)
```

## Label Encoding

```python
# Label encode 'sex'
le = LabelEncoder()
df['sex_le'] = le.fit_transform(df['sex'])
df[['sex', 'sex_le']].drop_duplicates()
```

## Ordinal Encoding

```python
# Define order for 'class'
ordinal_order = [['Third', 'Second', 'First']]
oe = OrdinalEncoder(categories=ordinal_order)
df['class_ordinal'] = oe.fit_transform(df[['class']])
df[['class', 'class_ordinal']].drop_duplicates()
```

## One-Hot Encoding

```python
# One-hot encode 'embarked'
df_ohe = pd.get_dummies(df, columns=['embarked'], prefix='embarked')
#df_ohe.head()
df_ohe[['embarked_C', 'embarked_Q', 'embarked_S']].drop_duplicates()
df[['embarked']].drop_duplicates()
```
