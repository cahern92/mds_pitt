```python
df = pd.read_csv("APMM9-dataset10.csv")
```

```python
df.head() 
```

```python
from sklearn.metrics import confusion_matrix, precision_score, recall_score, f1_score
```

```python
# For Label 1
df["act_lab1"] = np.where(df.actual == "label1", 1, 0)
df["pred_lab1"] = np.where(df.predicted == "label1", 1, 0)
```

```python
df
```

```python
TN, FP, FN, TP = confusion_matrix(df.act_lab1, df.pred_lab1).ravel()
```

```python
precision = precision_score(df.act_lab1, df.pred_lab1)
recall = recall_score(df.act_lab1, df.pred_lab1)
f1 = f1_score(df.act_lab1, df.pred_lab1)
```

```python
print(f"TP: {TP}")
print(f"FP: {FP}")
print(f"TN: {TN}")
print(f"FN: {FN}")
print(f"(Precision: {precision}")
print(f"(Recall: {recall}")
print(f"(F1 Score: {f1}")
```
