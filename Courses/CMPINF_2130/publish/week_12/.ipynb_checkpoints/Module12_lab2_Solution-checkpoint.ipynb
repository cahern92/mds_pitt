```python
import geodatasets
import geopandas as gpd
import folium
from shapely.geometry import Point, LineString, Polygon

```

```python
#use airbnb data to get the data for the city of chicago, you can access it uner geoda
#your code goes here
chicago = gpd.read_file(geodatasets.get_url("geoda airbnb"))

# Create a Folium map
# pick a right level of zoom
#your code goes here
m = folium.Map(location=[41.87834849785068, -87.64587372637078], zoom_start=10)

# Iterate through the Chicago neighbohroods and add a marker for neighborhood with population larget than 50,000
# For each marker add a popup text that shows the neighborhood name and population
#your code goes here
for idx, row in chicago.iterrows():
    centroid = row.geometry.centroid  # Get centroid of neighborhood
    community_name = row["community"]
    population = int(row["population"])

    # Create marker popup text
    popup_text = f"""
    Community: {community_name} <br>
    population: {population} <br>
    """
    if (population > 50000):
        # Add marker to map
        folium.Marker(
            location=[centroid.y, centroid.x],  # (lat, lon)
            popup=popup_text,
            icon=folium.Icon(color="blue", icon="info-sign")
        ).add_to(m)

m
```
