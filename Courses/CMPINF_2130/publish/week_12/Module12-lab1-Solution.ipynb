```python
import geodatasets
import geopandas as gpd

# Explore datasets available at geodatasets
geodatasets.data
```

```python
# Use Airbnb data to get the data for the city of Chicago, you can access it using geoda 

chicago = gpd.read_file(geodatasets.get_url("geoda airbnb"))
```

```python
# Explore the data and fields included in the data using the "head" function

print(chicago.head())
```

```python
# Plot the map of the city of Chicago

chicago.plot(figsize=(10, 6), edgecolor='red')
```

```python
# Filter to include only the neighborhood of Oakland in Chicago and plot that neighborhood

oakland = chicago[chicago['community']=='OAKLAND']
oakland.plot(figsize=(8,6), edgecolor='black',color='lightblue')
```
