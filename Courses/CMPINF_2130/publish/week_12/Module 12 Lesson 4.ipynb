### Import Modules

```python
import geodatasets
import geopandas as gpd
import folium
from shapely.geometry import Point, LineString, Polygon
import pandas as pd
```

```python
# Create basic map
m = folium.Map(location=[37.7749, -122.4194], zoom_start=10)
m
```

```python
# Add a marker and popup information
folium.Marker(
    location=[37.7749, -122.4194], # Latitude and Longitude of SF
    popup="San Francisco, CA", # Message when clicked
    tooltip="Click Me!", # Message on hover
    icon=folium.Icon(color="blue", icon="info-sign") # Custom icon
).add_to(m)
m
```

```python
# Add diff map tiles (satellite, terrain, etc.)
folium.TileLayer("OpenTopoMap", attr="test").add_to(m)
folium.TileLayer("Stadia.Outdoors", attr="test").add_to(m)
folium.TileLayer("Stadia.StamenTonerLife", attr="test").add_to(m)

# Add layer control to switch between maps
folium.LayerControl().add_to(m)
```
