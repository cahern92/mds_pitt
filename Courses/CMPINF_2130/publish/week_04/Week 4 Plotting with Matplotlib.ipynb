### Using Matplotlib
Using matplotlib to set up your code

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import seaborn as sns
```

```python
x_1 = np.linspace(0,5)
y_1 = x_1**2
```

```python
plt.plot(x_1, y_1)

plt.title('Days squared chart')
plt.xlabel('Days')
plt.ylabel('Days Squared')
```

Creating subplots aka multiple plots

```python
plt.subplot(1,2,1)
plt.plot(x_1, y_1, 'r')

plt.subplot(1,2,2)
plt.plot(x_1, y_1, 'b')
```
